## Question 1: Describe different types of data sources used in ETL with suitable examples.

## Answer

ETL (Extract, Transform, Load) processes can collect data from different types of sources:

1. **Relational Databases:** Store structured data in tables. Examples include MySQL, PostgreSQL, Oracle, and SQL Server.
2. **CSV/Flat Files:** Simple text files containing tabular data. Example: `sales.csv`.
3. **Excel Files:** Spreadsheet files used for reports and business records. Examples include `.xlsx` and `.xls`.
4. **APIs:** Allow applications to request data from external services. Example: a weather API.
5. **Cloud Storage:** Services such as Amazon S3, Google Cloud Storage, and Azure Blob Storage can store large datasets.
6. **NoSQL Databases:** Used for semi-structured or unstructured data. Examples include MongoDB and Cassandra.

**Example:** An e-commerce company may extract customer data from MySQL, sales data from CSV files, and product information through an API.

## Question 2: What is data extraction? Explain its role in the ETL pipeline.

## Answer

**Data extraction** is the process of collecting data from one or more source systems so it can be processed and stored in a target system.

It is the first stage of the ETL pipeline:

**Extract → Transform → Load**

During extraction, an ETL system connects to sources such as databases, CSV files, Excel files, APIs, or cloud storage and retrieves the required data.

The main roles of extraction are:
- Collecting data from different sources.
- Selecting the required records and columns.
- Supporting full or incremental extraction.
- Making source data available for transformation.
- Maintaining the accuracy and consistency of extracted data.

## Question 4: Explain the steps involved in extracting data from a relational database.

## Answer

The typical steps are:

1. Identify the database and required tables.
2. Connect to the database using the appropriate driver and credentials.
3. Understand the schema, columns, keys, and relationships.
4. Write a SQL query for the required data.
5. Apply filters to avoid unnecessary records.
6. Choose full or incremental extraction.
7. Retrieve large datasets in batches.
8. Validate row counts, data types, nulls, and duplicates.
9. Store the extracted data in a staging area.

## Question 5: Explain three common challenges faced during data extraction.

## Answer

### 1. Large Data Volume
Millions of records can consume considerable memory, storage, bandwidth, and processing time.

**Solution:** Use batching, pagination, incremental extraction, and compression.

### 2. Data Quality Problems
Source data can contain missing values, duplicate records, invalid formats, and inconsistent values.

**Solution:** Perform validation and data-quality checks.

### 3. Source Performance and Availability
Heavy extraction queries can affect production systems. Network problems or source-system downtime can also interrupt extraction.

**Solution:** Optimize queries, use suitable indexes or read replicas, schedule extraction carefully, and implement retries and checkpoints.

## Question 6: What are APIs? Explain how APIs help in real-time data extraction.

## Answer

**API (Application Programming Interface)** is a method that allows software applications to communicate and exchange data.

APIs help with real-time or near-real-time extraction because an ETL system can request the latest available information from an external service without directly accessing its database.

**Examples:**
- Weather APIs for current weather information.
- Payment APIs for transaction information.
- E-commerce APIs for product and order information.

A typical API extraction process is:

**Send Request → Authenticate → Receive Response → Parse JSON/XML → Validate → Transform → Load**

Developers should handle authentication, pagination, rate limits, timeouts, retries, and API errors.

## Question 7: Why are databases preferred for enterprise-level data extraction?

## Answer

Databases are preferred in enterprise environments because they are designed to manage large amounts of data reliably and efficiently.

Major advantages include:

- **Scalability:** Can handle large datasets.
- **Performance:** Indexes and query optimization improve data retrieval.
- **Security:** Authentication, roles, and permissions protect data.
- **Data Integrity:** Constraints and transactions help maintain consistency.
- **Concurrent Access:** Multiple users and applications can access data safely.
- **Reliability:** Backup, recovery, replication, and high-availability features are available.
- **Structured Querying:** SQL supports filtering, joining, sorting, and aggregation.
- **Incremental Extraction:** Timestamps, change tracking, or CDC can help extract only changed data.

Because of these features, enterprise ETL pipelines commonly extract data from operational databases and load it into data warehouses or data lakes.

## Question 8: What steps should an ETL developer take when extracting data from large CSV files (1GB+)?

## Answer

For CSV files larger than 1GB, the complete file should generally not be loaded into memory at once.

Recommended steps:

1. Inspect the delimiter, header, encoding, columns, and data types.
2. Use chunk-based or streaming reads.
3. Validate each chunk for malformed rows, missing values, duplicates, and incorrect data types.
4. Process data incrementally instead of keeping the complete dataset in RAM.
5. Write output in batches to the target or staging system.
6. Use compression when appropriate.
7. Track progress with checkpoints so failed jobs can resume.
8. Use efficient parsing tools and avoid unnecessary operations.
9. Monitor memory, CPU, disk, and network usage.
10. Validate final output using row counts and data-quality checks.
